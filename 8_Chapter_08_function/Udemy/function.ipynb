```python
# variable length arguement 

## 1 Positional Arguement (*)

def print_numbers(*args):           #  *args is not mandatory use ...we can also *any_name
    for num in args:
        print(num,",",end=" ")

print_numbers(1,2,3,5.4,9,"orvision",True)
```

```python
## 2 Keyword Arguement  (**)
#  -->all parameters are in the  'key, value'  pair

def keyword(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

keyword(name="onkar",roll_no=21,Grade="A",CGPA=8.90)
```

```python
# bothb combiine 

def keyword(*args,**kwargs):
    for num in args:
        print("Positional arg: ",num)
    for key,value in kwargs.items():
        print(f"{key}:{value}")

keyword(1,23,75,"sam",name="onkar",roll_no=21,Grade="A",CGPA=8.90)
```
